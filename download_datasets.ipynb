```python
import os, io, zipfile, requests
import numpy as np

BASE = "http://cs.uef.fi/sipu/datasets/"
DEST = ".datasets"
os.makedirs(DEST, exist_ok=True)

def fetch(url_name, save_as):
    path = os.path.join(DEST, save_as)
    if not os.path.exists(path):
        r = requests.get(BASE + url_name, timeout=120)
        r.raise_for_status()
        open(path, "wb").write(r.content)
    return path

shape_files = {
    "Aggregation.txt": "Aggregation.txt", "Compound.txt": "Compound.txt",
    "D31.txt": "D31.txt", "R15.txt": "R15.txt", "flame.txt": "flame.txt",
    "jain.txt": "jain.txt", "pathbased.txt": "pathbased.txt", "spiral.txt": "spiral.txt",
    "Overlap.txt": "overlap.txt", "overlap.pa": "overlap.pa",
    "Skewed.txt": "skewed.txt", "skewed.pa": "skewed.pa",
    "Asymmetric.txt": "asymmetric.txt", "asymmetric.txt": "asymmetric.txt", "asymmetric.pa": "asymmetric.pa",
    "s1.txt": "s1.txt", "s2.txt": "s2.txt", "s3.txt": "s3.txt", "s4.txt": "s4.txt",
}
for save_as, url_name in shape_files.items():
    fetch(url_name, save_as)
print(f"downloaded {len(shape_files)} shape files")
```

```python
zbytes = requests.get(BASE + "s-originals.zip", timeout=120).content
z = zipfile.ZipFile(io.BytesIO(zbytes))
for i in (1, 2, 3, 4):
    open(os.path.join(DEST, f"s{i}-label.pa"), "wb").write(z.read(f"s{i}-label.pa"))
print("downloaded S-set labels")
```

```python
fetch("unbalance2.txt", "Unbalance2.txt")
fetch("unbalance2.pa", "Unbalance2.pa")
print("downloaded Unbalance2")
```
